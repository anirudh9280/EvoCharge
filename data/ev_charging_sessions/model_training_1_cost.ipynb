```python
from sklearn.impute import SimpleImputer 
from sklearn.compose import ColumnTransformer 
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder 
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error 
from sklearn.model_selection import train_test_split
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv('ev_charging_patterns.csv')

df['start_time'] = pd.to_datetime(df['Charging Start Time'])
df['hour_start'] = df['start_time'].dt.hour
df['start_date'] = df['start_time'].dt.date

X = df.copy() 
X.dropna(subset='Charging Cost (USD)', axis=0, inplace=True)
y = X['Charging Cost (USD)']
X.drop('Charging Cost (USD)', axis=1, inplace=True)
X.drop('Charging Start Time', axis=1, inplace=True)
X.drop('Charging End Time', axis=1, inplace=True)

X = X[['hour_start', 'Vehicle Model', 'Battery Capacity (kWh)', 'Charging Duration (hours)',
'Time of Day', 'Day of Week', 'State of Charge (Start %)', 'Temperature (°C)', 'Vehicle Age (years)', 'Charger Type']] 

X.head()
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(X, y, train_size=0.8, test_size=0.2, random_state=0)

categorical_cols = [col for col in X_train.columns if X_train[col].dtype in ['category', 'object']] # if X_train[col].nunique() < 10 and 

numerical_cols = [col for col in X_train.columns if X_train[col].dtype in ['int64', 'float64']]

my_cols = categorical_cols + numerical_cols #only keep int64, float64, and object columns
X_train = X_train[my_cols].copy()
X_valid = X_valid[my_cols].copy()

#preprocessing for NUMERICAL/CATEGORICAL data (both use SimpleImputer to replace null values with mean)
numerical_transformer = SimpleImputer(strategy='constant') #strategy="constant" for numerical
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')), #strategy="most_frequent" for categorical
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])
```

```python
#bundle preprocessing for numerical and categorical data (applies n_transformer to n_cols and c_transformer to c_cols)
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_cols),
        ('cat', categorical_transformer, categorical_cols)
    ])
model = RandomForestRegressor(n_estimators=100, random_state=0) #min_samples_split, max_depth, criterion
```

```python
clf = Pipeline(steps=[('preprocessor', preprocessor), 
                      ('model', model)])

#preprocessing of training data, fit model (USE PIPELINE TO FIT, NOT MODEL)
clf.fit(X_train, y_train)
```

```python
preds = clf.predict(X_valid)
print('MAE:', mean_absolute_error(y_valid, preds))
```
